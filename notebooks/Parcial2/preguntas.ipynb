# Capítulo 7: linked list

```python
class SinglyLinkedList:
    def __init__(self):
        self._head = None
        self._tail = None
        self._size = 0

    class _Node:
        __slots__ = '_element', '_next'
        def __init__(self, element, nxt):
            self._element = element
            self._next    = nxt

    def __len__(self):
        return self._size
    
    def is_empty(self):
        return self._size == 0
    
    def first(self):
        return self._head._element

    def insert_first(self, e):
        self._head = self._Node(e, self._head)
        if self._size == 0:
            self._tail = self._head
        self._size += 1

    def insert_last(self, e):
        node = self._Node(e, None)
        if self.is_empty():
            self._head = node
        else:
            self._tail._next = node
        self._tail = node
        self._size += 1

    def remove_first(self):
        val = self._head._element
        self._head = self._head._next
        self._size -= 1
        if self._size == 0:
            self._tail = None
        return val

S = SinglyLinkedList()
for x in ['A', 'B', 'C', 'D']:
    S.insert_last(x)
print(len(S), S.first())    # 4  A
S.remove_first()
print(len(S), S.first())    # 3  B
```

```python
def build(items):
    """Construye una SinglyLinkedList con los elementos dados."""
    L = SinglyLinkedList()
    for x in items:
        L.insert_last(x)
    return L

def to_list(L):
    """Recorre L desde la cabeza y retorna sus elementos en una lista de Python."""
    out, node = [], L._head
    while node is not None:
        out.append(node._element)
        node = node._next
    return out
```

```python
from goodrich.ch08.linked_binary_tree import LinkedBinaryTree
from goodrich.ch08.expression_tree import tokenize, build_expression_tree

def make_tree(spec):
    """Construye un LinkedBinaryTree a partir de tuplas anidadas (elemento, izq, der).
    Una hoja puede escribirse solo con su elemento y un hijo ausente con None.
    Ejemplo: ('A', ('B', 'D', None), 'C')"""
    T = LinkedBinaryTree()
    if spec is not None:
        if not isinstance(spec, tuple):
            spec = (spec, None, None)
        _fill(T, T._add_root(spec[0]), spec)
    return T

def _fill(T, p, spec):
    _, left, right = spec
    if left is not None:
        if not isinstance(left, tuple):
            left = (left, None, None)
        _fill(T, T._add_left(p, left[0]), left)
    if right is not None:
        if not isinstance(right, tuple):
            right = (right, None, None)
        _fill(T, T._add_right(p, right[0]), right)

def elements(positions):
    """Lista con los elementos de una secuencia de posiciones (p. ej. T.preorder())."""
    return [p.element() for p in positions]
```

```python
import random
from math import ceil, log2
from itertools import permutations
from goodrich.ch09.heap_priority_queue import HeapPriorityQueue
from goodrich.ch09.adaptable_heap_priority_queue import AdaptableHeapPriorityQueue
from goodrich.exceptions import Empty

def is_heap(keys):
    """True si la lista de claves (representación por arreglo) cumple el orden de heap mínimo."""
    return all(keys[(j - 1) // 2] <= keys[j] for j in range(1, len(keys)))

def heap_keys(H):
    """Claves del heap H en el orden del arreglo interno."""
    return [item._key for item in H._data]

import hashlib

def huella(x):
    """Huella (hash) de una respuesta: permite verificarla sin mostrarla."""
    return hashlib.sha256(repr(x).encode()).hexdigest()[:16]

def possible_indices(rank, n):
    """Búsqueda exhaustiva: índices del arreglo en los que puede quedar la clave de ese rango
    (1 = la menor) en algún heap con las claves distintas 1..n. Úsala para comprobar ideas."""
    return {perm.index(rank) for perm in permutations(range(1, n + 1)) if is_heap(perm)}
```
