## 7.1 Lista Enlazada Simple (*Singly Linked List*)

Una **lista enlazada simple** almacena una secuencia de nodos, donde cada nodo guarda:

- Un **elemento** (`_element`): el dato almacenado.
- Una **referencia al siguiente** (`_next`): enlace al nodo sucesor (`None` en el último).

La lista mantiene solo una referencia al primer nodo (`head`). Opcionalmente mantiene `tail` para insertar al final en $O(1)$.

**ADT — operaciones y complejidad:**

| Operación | Descripción | $O$ |
|---|---|---|
| `insert_first(e)` | Inserta al frente | $O(1)$ |
| `insert_last(e)` | Inserta al final (con `tail`) | $O(1)$ |
| `remove_first()` | Elimina y retorna el primer elemento | $O(1)$ |
| `remove_last()` | Requiere recorrer hasta el penúltimo | $O(n)$ |
| `first()` | Retorna el primer elemento sin eliminarlo | $O(1)$ |
| Búsqueda / acceso por índice | Recorrido desde `head` | $O(n)$ |

**Clave:** inserción y eliminación al **frente** son $O(1)$ — no hay desplazamiento de elementos como en el arreglo.

![Lista Enlazada Simple](../assets/ch07_singly_linked.png)

````python
class _Node:
    __slots__ = '_element', '_next'
    def __init__(self, element, nxt):
        self._element = element
        self._next    = nxt

class SinglyLinkedList:
    def __init__(self):
        self._head = None
        self._tail = None
        self._size = 0

    def __len__(self):    return self._size
    def is_empty(self):   return self._size == 0
    def first(self):      return self._head._element

    def insert_first(self, e):
        self._head = _Node(e, self._head)
        if self._size == 0:
            self._tail = self._head
        self._size += 1

    def insert_last(self, e):
        node = _Node(e, None)
        if self.is_empty():
            self._head = node
        else:
            self._tail._next = node
        self._tail = node
        self._size += 1

    def remove_first(self):
        val = self._head._element
        self._head = self._head._next
        self._size -= 1
        if self._size == 0:
            self._tail = None
        return val

S = SinglyLinkedList()
for x in ['A', 'B', 'C', 'D']:
    S.insert_last(x)
print(len(S), S.first())    # 4  A
S.remove_first()
print(len(S), S.first())    # 3  B
````

## 7.2 Lista Enlazada Circular (*Circularly Linked List*)

Igual que la simple, pero el **`_next` del último nodo apunta de regreso al primero**: forma un ciclo. Se mantiene solo la referencia **`_tail`** (la cola).

- **Frente**: `_tail._next`
- **Rotar** (`rotate`): `_tail = _tail._next` — mueve el frente al final en $O(1)$

**ADT CircularQueue:**

| Operación | Descripción | $O$ |
|---|---|---|
| `enqueue(e)` | Inserta después de `tail`, avanza `tail` | $O(1)$ |
| `dequeue()` | Elimina `tail._next` (el frente) | $O(1)$ |
| `rotate()` | El frente pasa al final | $O(1)$ |
| `first()` | `tail._next._element` | $O(1)$ |

**Aplicación típica:** planificación *Round-Robin* — cada proceso recibe su turno y regresa al final.

## 7.3 Lista Doblemente Enlazada (*Doubly Linked List*)

Cada nodo tiene **tres campos**: referencia al nodo anterior (`_prev`), el elemento, y referencia al siguiente (`_next`). Esto permite:

- Recorrer la lista en **ambas direcciones**.
- **Eliminar cualquier nodo en $O(1)$** dada su referencia directa.

### Nodos centinela (*sentinels*)

Goodrich usa dos nodos especiales vacíos en los extremos:
- **`_header`**: antes del primer elemento real — su `_prev` es `None`.
- **`_trailer`**: después del último elemento real — su `_next` es `None`.

Los centinelas **eliminan casos borde**: nunca hay que preguntar "¿es el primero?" o "¿es el último?" al insertar o eliminar.

**Invariante siempre válido (lista con elementos A, B, C):**
```
_header ↔ A ↔ B ↔ C ↔ _trailer
```
Si la lista está vacía: `_header._next = _trailer` y `_trailer._prev = _header`.

**Operaciones (todas $O(1)$ dado el nodo):**

| Operación interna | Descripción |
|---|---|
| `_insert_between(e, pred, succ)` | Inserta `e` entre dos nodos conocidos |
| `_delete_node(node)` | Elimina un nodo dado |
| Insertar al frente | `_insert_between(e, _header, _header._next)` |
| Insertar al final | `_insert_between(e, _trailer._prev, _trailer)` |

![Lista Doblemente Enlazada con Centinelas](../assets/ch07_doubly_linked.png)

````python
from goodrich.ch07.linked_deque import *
d = LinkedDeque()
for x in [1, 2, 3]: d.insert_last(x)
d.insert_first(0)
print(d.first(), d.last(), len(d))   # 0  3  4
d.delete_first(); d.delete_last()
print(d.first(), d.last(), len(d))   # 1  2  2
````

## 7.4 ADT Lista Posicional (*Positional List*)

La **lista posicional** envuelve una lista doblemente enlazada y expone objetos **`Position`** como referencias estables a nodos. A diferencia de un índice (que cambia al insertar/eliminar), una posición sigue siendo válida mientras el nodo exista.

**Operaciones principales — todas $O(1)$:**

| Operación | Descripción |
|---|---|
| `L.first()` / `L.last()` | Primera / última posición (`None` si vacía) |
| `L.before(p)` / `L.after(p)` | Posición anterior / siguiente a `p` |
| `L.add_first(e)` / `L.add_last(e)` | Inserta en los extremos |
| `L.add_before(p, e)` / `L.add_after(p, e)` | Inserta junto a la posición `p` |
| `L.replace(p, e)` | Reemplaza el elemento en la posición `p` |
| `L.delete(p)` | Elimina la posición `p` y retorna su elemento |

**Iteración** `for e in L` recorre en orden: $O(n)$.

**Caso de uso típico:** mantener listas ordenadas con frecuentes inserciones/eliminaciones en posiciones intermedias (p.ej., lista de favoritos con acceso frecuente — §7.6 en Goodrich).

## Comparación: Arreglo vs Listas Enlazadas

| Operación | Array dinámico | Lista simple | Lista doble |
|---|---|---|---|
| Acceso por índice `k` | $O(1)$ | $O(n)$ | $O(n)$ |
| Insertar / eliminar al frente | $O(n)$ | $O(1)$ | $O(1)$ |
| Insertar / eliminar al final | $O(1)$* | $O(1)$† | $O(1)$ |
| Insertar / eliminar en posición media | $O(n)$ | $O(n)$ | $O(1)$‡ |
| Memoria por elemento | Compacta | Nodo + 1 puntero | Nodo + 2 punteros |

\* amortizado — † requiere mantener `tail` — ‡ requiere referencia directa al nodo

**Regla práctica:**
- Muchos accesos por índice → **arreglo**.
- Muchas inserciones/eliminaciones al frente o en posición conocida → **lista enlazada**.
