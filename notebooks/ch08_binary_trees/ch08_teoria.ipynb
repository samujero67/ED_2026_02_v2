## 8.1 Árbol general — definición

- Un **árbol** es una colección de nodos con una relación padre-hijo.
- La **raíz** es el único nodo sin padre.
- Cada nodo (salvo la raíz) tiene **exactamente un padre** y puede tener **cualquier número de hijos** (0, 1, 2, ...).
- Una **hoja** es un nodo sin hijos. Un **nodo interno** tiene al menos un hijo.
- **Profundidad** de un nodo: número de aristas desde la raíz hasta ese nodo.
- **Altura** de un árbol: la profundidad máxima entre todas sus hojas.

**Aplicación típica:** jerarquías del mundo real — un catálogo de productos (categoría → subcategoría → producto), un sistema de archivos (carpeta → subcarpeta → archivo), un organigrama de una empresa. En todos estos casos un nodo puede tener muchos hijos, no solo dos.

## 8.2 Árbol binario — definición

- Un **árbol binario** es un árbol donde cada nodo tiene **como máximo 2 hijos**.
- Los hijos se distinguen como **hijo izquierdo** e **hijo derecho** (importa el orden).
- Es un caso particular de árbol general: todo lo dicho arriba (raíz, hoja, profundidad, altura) aplica igual.

### ADT — Árbol binario

Un **Tipo Abstracto de Dato (ADT)** define *qué* operaciones ofrece la estructura, sin decir *cómo* están implementadas por dentro. Cada nodo se referencia mediante una **posición** (`p`).

| Método | Descripción | $O$ (representación enlazada) |
|---|---|---|
| `root()` | Posición de la raíz (`None` si el árbol está vacío) | $O(1)$ |
| `parent(p)` | Posición del padre de `p` (`None` si `p` es la raíz) | $O(1)$ |
| `left(p)` | Posición del hijo izquierdo de `p` (`None` si no existe) | $O(1)$ |
| `right(p)` | Posición del hijo derecho de `p` (`None` si no existe) | $O(1)$ |
| `sibling(p)` | Posición del hermano de `p` (`None` si no tiene) | $O(1)$ |
| `children(p)` | Genera las posiciones de los hijos de `p` | $O(c_p)$, con $c_p \le 2$ |
| `num_children(p)` | Número de hijos de `p` (0, 1 o 2) | $O(1)$ |
| `is_root(p)` | `True` si `p` es la raíz | $O(1)$ |
| `is_leaf(p)` | `True` si `p` no tiene hijos | $O(1)$ |
| `is_empty()` | `True` si el árbol no tiene nodos | $O(1)$ |
| `len(T)` | Número total de nodos del árbol | $O(1)$ |
| `iter(T)` | Itera sobre los **elementos** del árbol (recorriendo todas las posiciones) | $O(n)$ |

**Notas:**
- `depth(p)` y `height(p)` (número de aristas hasta la raíz / hasta la hoja más profunda del subárbol) suelen agregarse como operaciones derivadas del ADT, ya que se calculan a partir de `parent`, `left` y `right`.
- El ADT no incluye recorridos como método aparte: `preorder`, `inorder`, `postorder` y `bfs` son formas de **generar** el orden en que se visitan las posiciones — se apoyan en el ADT, no lo modifican.
- El ADT tampoco fija cómo insertar o eliminar nodos: eso depende de para qué se use el árbol (p. ej. un árbol de búsqueda binaria inserta según el valor; un árbol de expresión se construye de una vez a partir de la expresión).

## 8.3 Recorridos

- **Preorden** (Raíz → Izq → Der): visita el nodo antes que sus subárboles.
- **Inorden** (Izq → Raíz → Der): visita el subárbol izquierdo, luego el nodo, luego el derecho. Solo tiene sentido en árboles **binarios** (necesita distinguir izquierda de derecha).
- **Postorden** (Izq → Der → Raíz): visita ambos subárboles antes que el nodo.
- **BFS / por niveles**: visita los nodos nivel por nivel, usando una cola.

Preorden, postorden y BFS también se usan en árboles **generales** (con cualquier número de hijos); inorden es propio de árboles binarios.

## 8.4 Aplicación: árbol de expresiones aritméticas

Una expresión aritmética se puede representar como un árbol binario:

- Cada **nodo interno** es un operador (`+`, `-`, `×`, `/`).
- Cada **hoja** es un operando (un número).
- Los hijos de un operador son sus dos operandos (izquierdo y derecho).

Los recorridos de la sección anterior, aplicados a este árbol, **generan las tres notaciones clásicas** de una expresión:

| Recorrido | Notación | Resultado sobre el árbol de abajo |
|---|---|---|
| Preorden | Prefija (polaca) | `- / × + 3 1 3 + - 9 5 2 + × 3 - 7 4 6` |
| Inorden | Infija (la que escribimos normalmente, con paréntesis) | `((((3+1)×3)/((9-5)+2)) - ((3×(7-4))+6))` |
| Postorden | Postfija (polaca inversa) | `3 1 + 3 × 9 5 - 2 + / 3 7 4 - × 6 + -` |

**Por qué importa la postfija:** en notación postfija no se necesitan paréntesis ni reglas de precedencia — se puede **evaluar** la expresión con una sola pasada usando una pila (se apilan operandos; al encontrar un operador, se desapilan dos operandos, se opera y se vuelve a apilar el resultado). Es la misma idea que usan las calculadoras y los intérpretes de expresiones.

## 8.5 Para la clase: implementa el ADT

Ahora implementa el ADT de la sección 8.2 con **nodos enlazados directos** — igual que `SinglyLinkedList` en CH07, pero con el nodo (`_Node`) **anidado dentro de la clase del árbol** (como lo hace Goodrich): no se usa desde afuera, así que no necesita vivir a nivel de módulo. **Sin** clase `Position` ni `_validate`: el nodo mismo es lo que se pasa y retorna. Los nombres de los métodos se mantienen en inglés, igual que en el ADT de la sección 8.2.

Completa cada método marcado con `pass`.

**Por qué `add_root`, `add_left` y `add_right` lanzan `ValueError` en vez de sobreescribir:** un árbol binario tiene una sola raíz y cada nodo tiene a lo más un hijo izquierdo y uno derecho. Si se permitiera reemplazar sin avisar, el subárbol que colgaba de ahí queda **huérfano** — sigue en memoria pero ya nada lo referencia — y el contador `_size` queda desincronizado del árbol real. El `raise` convierte ese error de programación en un aviso inmediato, en vez de un bug silencioso.

```python
class LinkedBinaryTree:
    """Árbol binario con nodos enlazados directos (sin Position)."""

    class _Node:
        """Nodo interno: elemento, padre, hijo izquierdo, hijo derecho."""
        __slots__ = ('_element', '_parent', '_left', '_right')

        def __init__(self, element, parent=None, left=None, right=None):
            self._element = element
            self._parent = parent
            self._left = left
            self._right = right

        def __repr__(self):
            return f'Node({self._element})'

    def __init__(self):
        self._root = None
        self._size = 0

    def __len__(self):
        pass

    def is_empty(self):
        pass

    def root(self):
        """Retorna el nodo raíz (None si el árbol está vacío)."""
        pass

    def parent(self, node):
        pass

    def left(self, node):
        pass

    def right(self, node):
        pass

    def num_children(self, node):
        pass

    def is_root(self, node):
        pass # debe usar `is`

    def is_leaf(self, node):
        pass

    def sibling(self, node):
        pass

    def children(self, node):
        """Genera los hijos de node (izquierdo y/o derecho, si existen)."""
        pass

    # --- construcción (ya implementadas, para poder armar árboles de prueba) ---
    def add_root(self, e):
        if self._root is not None:
            raise ValueError('La raíz ya existe')
        self._root = self._Node(e)
        self._size = 1
        return self._root

    def add_left(self, node, e):
        if node._left is not None:
            raise ValueError('Ya existe hijo izquierdo')
        node._left = self._Node(e, parent=node)
        self._size += 1
        return node._left

    def add_right(self, node, e):
        if node._right is not None:
            raise ValueError('Ya existe hijo derecho')
        node._right = self._Node(e, parent=node)
        self._size += 1
        return node._right
```

```python
# --- Tests: árbol A-G, el mismo del dibujo de la sección 8.2 ---
T = LinkedBinaryTree()
A = T.add_root('A')
B = T.add_left(A, 'B')
C = T.add_right(A, 'C')
D = T.add_left(B, 'D')
E = T.add_right(B, 'E')
F = T.add_left(C, 'F')
G = T.add_right(C, 'G')

print(len(T))                                   # 7
print(T.root() is A)                            # True
print(T.parent(D) is B)                         # True
print(T.left(A) is B, T.right(A) is C)          # True True
print(T.sibling(D) is E)                        # True
print(T.num_children(C))                        # 2
print(T.is_leaf(D), T.is_root(A))               # True True
print([n._element for n in T.children(A)])      # ['B', 'C']
```
