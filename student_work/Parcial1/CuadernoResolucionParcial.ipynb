```python
def MinMax(L):
    # Caso Base
    if len(L) == 1:
        return L[0], L[0]
    
    # Paso Recursivo
    max_resto, min_resto = MinMax(L[1:])
    
    # Evaluar máximo
    if L[0] > max_resto:
        max_final = L[0]
    else:
        max_final = max_resto
        
    # Evaluar mínimo
    if L[0] < min_resto:
        min_final = L[0]
    else:
        min_final = min_resto
        
    return max_final, min_final
```

```python
L=[1,0,7,0,2,-7]
MinMax(L)
```

Version Iterativa

```python
def Picos(L):
    if len(L) < 3:
        return 0
    
    total_picos = 0
    
    for i in range(1, len(L) - 1):
        actual = L[i]
        izquierda = L[i - 1]
        derecha = L[i + 1]
        
        if actual > izquierda and actual > derecha:
            total_picos = total_picos + 1
            
    return total_picos
```

```python
L=[1,0,7,0,2,-23]
Picos(L)
```

Version Recursiva

```python
def Picos(L):
    if len(L) < 3:
        return 0

    izquierda = L[0]
    centro = L[1]
    derecha = L[2]
    
    if centro > izquierda and centro > derecha:
        pico_e = 1
    else:
        pico_e = 0
    return pico_e + Picos(L[1:])
```

```python
L=[1,0,7,0,2,-23]
Picos(L)
```
