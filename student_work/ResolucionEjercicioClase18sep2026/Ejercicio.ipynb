# Queue

```python
class QueueSPQ:
       
    def __init__(self):
        """Crea una nueva cola vacía."""
        self._spq = SortedPriorityQueue()
        self._contador = 0  # Clave ascendente para mantener el orden de llegada

    def __len__(self):
        """Retorna el número de elementos en la cola."""
        return len(self._spq)

    def is_empty(self):
        """Retorna True si la cola está vacía."""
        return len(self._spq) == 0

    def enqueue(self, elemento):
        """Agrega un elemento al final de la cola.
        
        Asignamos el contador actual como clave y luego lo incrementamos 
        para el próximo elemento.
        """
        self._spq.add(self._contador, elemento)
        self._contador += 1

    def dequeue(self):
        """Elimina y retorna el primer elemento del frente de la cola (FIFO).
        
        Lanza la excepción Empty si la cola está vacía.
        """
        if self.is_empty():
            raise ValueError('La cola está vacía.')
        _, valor = self._spq.remove_min()
        return valor

    def first(self):
        """Retorna (pero sin eliminar) el elemento al frente de la cola.
        
        Lanza la excepción Empty si la cola está vacía.
        """
        if self.is_empty():
            raise ValueError('La cola está vacía.')
        _, valor = self._spq.min()
        return valor
```

# Stack

```python
class StackSPQ:    

    def __init__(self):
        """Crea una nueva pila vacía."""
        self._spq = SortedPriorityQueue()
        self._contador = 0  # Clave decreciente para priorizar el último en entrar

    def __len__(self):
        """Retorna el número de elementos en la pila."""
        return len(self._spq)

    def is_empty(self):
        """Retorna True si la pila está vacía."""
        return len(self._spq) == 0

    def push(self, elemento):
        """Agrega un elemento a la parte superior (tope) de la pila.
        
        Decrementamos el contador para que el elemento más reciente tenga 
        la clave más pequeña (más negativa) y sea el próximo en salir.
        """
        self._contador -= 1
        self._spq.add(self._contador, elemento)

    def pop(self):
        """Elimina y retorna el elemento del tope de la pila (LIFO).
        
        Lanza la excepción Empty si la pila está vacía.
        """
        if self.is_empty():
            raise ValueError('La pila está vacía.')
        _, valor = self._spq.remove_min()
        return valor

    def top(self):
        """Retorna (pero sin eliminar) el elemento del tope de la pila.
        
        Lanza la excepción Empty si la pila está vacía.
        """
        if self.is_empty():
            raise ValueError('La pila está vacía.')
        _, valor = self._spq.min()
        return valor
```

```python
# --- PRUEBA DE LA PILA (LIFO) ---
pila = StackSPQ()
pila.push("Dato A")
pila.push("Dato B")
pila.push("Dato C")

print("--- Pila (LIFO) ---")
print("Tope:", pila.top())       # Salida: Dato C
print("Pop:", pila.pop())         # Salida: Dato C (el último en entrar)
print("Pop:", pila.pop())         # Salida: Dato B

# --- PRUEBA DE LA COLA (FIFO) ---
cola = QueueSPQ()
cola.enqueue("Cliente 1")
cola.enqueue("Cliente 2")
cola.enqueue("Cliente 3")

print("\n--- Cola (FIFO) ---")
print("Frente:", cola.first())    # Salida: Cliente 1
print("Dequeue:", cola.dequeue()) # Salida: Cliente 1 (el primero en entrar)
print("Dequeue:", cola.dequeue()) # Salida: Cliente 2
```

# Árbol Binario a partir de Lista

```python
class TreeFromList(LinkedBinaryTree):

    def __init__(self, values):
        super().__init__()
        values = values or []
        if values and values[0] is not None:
            root_pos = self._add_root(values[0])
            self._build(values, root_pos, 0)

    def _build(self, values, pos, idx):
        left_idx, right_idx = 2 * idx + 1, 2 * idx + 2

        if left_idx < len(values) and values[left_idx] is not None:
            left_pos = self._add_left(pos, values[left_idx])
            self._build(values, left_pos, left_idx)

        if right_idx < len(values) and values[right_idx] is not None:
            right_pos = self._add_right(pos, values[right_idx])
            self._build(values, right_pos, right_idx)
```
