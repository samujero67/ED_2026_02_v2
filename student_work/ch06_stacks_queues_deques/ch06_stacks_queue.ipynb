## Resumen del capítulo 6 — Stacks, Queues and Deques (Goodrich, Tamassia & Goldwasser)

### 6.1 Pilas (*Stacks*) — disciplina LIFO

Una **pila** almacena elementos de forma que el **último en entrar es el primero en salir** (*Last In, First Out*, LIFO). Solo se puede operar sobre un extremo, llamado la **cima** (*top*): se agregan y se quitan elementos únicamente ahí.

**Aplicaciones típicas:** historial de "deshacer" de un editor, la pila de llamadas de funciones de un programa, verificar que los paréntesis/llaves/corchetes de una expresión estén balanceados, el botón "atrás" del navegador, evaluación de expresiones postfijas.

<img src = '../assets/ch06_stack_vs_queue.png' width = '60%'>

### 6.1.1 Abstract Data Type: `Stack`

Un **ADT (Abstract Data Type)** especifica *qué* operaciones ofrece una estructura y su significado — no *cómo* están implementadas por dentro. Cualquier estructura que cumpla este contrato puede llamarse "una pila", sin importar si usa un arreglo o una lista enlazada.

| Operación | Efecto |
|---|---|
| `S.push(e)` | Agrega el elemento `e` a la cima de la pila |
| `S.pop()` | Remueve y retorna el elemento de la cima (LIFO). Lanza `Empty` si está vacía |
| `S.top()` | Retorna (sin remover) el elemento de la cima. Lanza `Empty` si está vacía |
| `S.is_empty()` | `True` si la pila no tiene elementos |
| `len(S)` | Número de elementos en la pila |

Todas las operaciones son **O(1)** (amortizado, si la implementación usa un arreglo dinámico por debajo).

### 6.2 Colas (*Queues*) — disciplina FIFO

Una **cola** almacena elementos de forma que el **primero en entrar es el primero en salir** (*First In, First Out*, FIFO). Los elementos entran por el **final** (`enqueue`) y salen por el **frente** (`dequeue`).

**Aplicaciones típicas:** cola de impresión, planificación de procesos (*round robin*), recorrido en anchura (BFS) de grafos y árboles, cualquier sistema de "el primero que llega, el primero que se atiende".

### 6.2.1 Abstract Data Type: `Queue`

| Operación | Efecto |
|---|---|
| `Q.enqueue(e)` | Agrega el elemento `e` al final de la cola |
| `Q.dequeue()` | Remueve y retorna el elemento del frente (FIFO). Lanza `Empty` si está vacía |
| `Q.first()` | Retorna (sin remover) el elemento del frente. Lanza `Empty` si está vacía |
| `Q.is_empty()` | `True` si la cola no tiene elementos |
| `len(Q)` | Número de elementos en la cola |

Todas las operaciones son **O(1)** (amortizado).

### 6.2.2 Implementación eficiente: arreglo circular

Si se implementara una cola con un arreglo "normal" —`dequeue` quitando el elemento `0` y recorriendo el resto una posición— cada `dequeue` costaría **O(n)** por el desplazamiento. La solución es un **arreglo circular**: se guarda un índice `_front` y se avanza con aritmética módulo la capacidad, en vez de mover físicamente los datos.

```
avail = (_front + _size) % capacidad     # dónde va el próximo enqueue
_front = (_front + 1) % capacidad         # a dónde se recorre el frente tras un dequeue
```

Cuando el arreglo se llena, se redimensiona igual que un arreglo dinámico (cap. 5): se duplica la capacidad y se copian los elementos **realineados** desde el índice 0.

### 6.3 Deques (*Double-Ended Queues*)

Un **deque** ("cola doblemente terminada") generaliza a la pila y a la cola: permite insertar y quitar elementos **por ambos extremos**. Con un deque se puede simular tanto una pila (usando siempre el mismo extremo) como una cola (insertando por un extremo y quitando por el otro).

**Aplicaciones típicas:** historial de navegación con "adelante" y "atrás", ventanas deslizantes (*sliding window*) en algoritmos, deshacer/rehacer. Python ya trae uno listo para usar: `collections.deque`.

### 6.3.1 Abstract Data Type: `Deque`

| Operación | Efecto |
|---|---|
| `D.add_first(e)` | Agrega `e` al frente del deque |
| `D.add_last(e)` | Agrega `e` al final del deque |
| `D.delete_first()` | Remueve y retorna el elemento del frente. Lanza `Empty` si está vacío |
| `D.delete_last()` | Remueve y retorna el elemento del final. Lanza `Empty` si está vacío |
| `D.first()` | Retorna (sin remover) el elemento del frente. Lanza `Empty` si está vacío |
| `D.last()` | Retorna (sin remover) el elemento del final. Lanza `Empty` si está vacío |
| `D.is_empty()` | `True` si el deque no tiene elementos |
| `len(D)` | Número de elementos en el deque |

Con una **lista doblemente enlazada** (cap. 7) las 8 operaciones son **O(1) en el peor caso** — a diferencia del arreglo circular, no hace falta amortizar nada porque no hay que redimensionar para insertar por el frente.

### 6.4 Eficiencia comparada

| Operación | `ArrayStack` | `ArrayQueue` (circular) | `LinkedDeque` |
|---|---|---|---|
| insertar/quitar en un extremo fijo | O(1) amortizado | O(1) amortizado | O(1) peor caso |
| `top` / `first` / `last` | O(1) | O(1) | O(1) |
| `is_empty`, `len` | O(1) | O(1) | O(1) |
| acceso a un elemento intermedio | no soportado por el ADT | no soportado por el ADT | no soportado por el ADT |

Ninguno de los tres ADT permite acceder a un elemento "de la mitad" — esa restricción es justamente lo que garantiza que todas sus operaciones sean O(1): siempre se opera sobre un extremo conocido.

### 6.5 Ideas clave para recordar

1. Un **ADT** es un contrato de operaciones (qué se puede hacer y qué garantiza), independiente de la implementación concreta (arreglo, arreglo circular, lista enlazada).
2. **Stack = LIFO**, un solo extremo activo (`top`). **Queue = FIFO**, dos extremos con roles fijos (`front` para salir, `back` para entrar). **Deque** = ambos extremos activos para entrar y salir.
3. Implementar una cola sobre un arreglo "plano" (sin circularidad) es ineficiente: quitar del frente obliga a desplazar todo lo demás, O(n) por operación.
4. El **arreglo circular** resuelve eso con aritmética módulo la capacidad — así `dequeue` es O(1) sin mover memoria.
5. Todas las operaciones válidas de estos tres ADT son O(1) (amortizado o peor caso, según la implementación) — si una operación necesita recorrer la estructura, esa operación **no pertenece** al ADT.

A continuación se definen `StackADT`, `QueueADT` y `DequeADT` como las clases que usted debe completar: cada `__init__` ya crea el arreglo (una lista de Python vacía), igual que hace `ArrayStack` en `goodrich/ch06`. El resto de los métodos está en blanco (solo con su descripción) para que usted implemente la lógica, siguiendo el contrato de las tablas de la sección anterior.

````python
class StackADT:
    """ADT de una Pila: disciplina LIFO (Last In, First Out)."""

    def __init__(self):
        self._data = []

    def __len__(self):
        """Número de elementos en la pila."""
        return len(self._data)

    def is_empty(self):
        """True si la pila no tiene elementos."""
        return len(self._data) == 0

    def push(self, e):
        """Agrega e a la cima de la pila."""
        self._data.append(e)

    def top(self):
        """Retorna (sin remover) el elemento de la cima. Lanza Empty si está vacía."""
        return self._data[-1]

    def pop(self):
        """Remueve y retorna el elemento de la cima. Lanza Empty si está vacía."""
        return self._data.pop()
````

````python
class QueueADT:
    """ADT de una Cola: disciplina FIFO (First In, First Out)."""

    def __init__(self):
        self._data = []

    def __len__(self):
        """Número de elementos en la cola."""
        return len(self._data)

    def is_empty(self):
        """True si la cola no tiene elementos."""
        return len(self) == 0

    def enqueue(self, e):
        """Agrega e al final de la cola."""
        self._data.append(e)

    def first(self):
        """Retorna (sin remover) el elemento del frente. Lanza Empty si está vacía."""
        return self._data[0]

    def dequeue(self):
        """Remueve y retorna el elemento del frente. Lanza Empty si está vacía."""
        return self._data.pop(0)
````

### ¿Por qué importa el arreglo circular? Comparación experimental

Se compara el costo de quitar repetidamente el **primer** elemento de una lista de Python (`list.pop(0)`, desplaza todo — O(n)) contra `collections.deque.popleft()` (implementación circular/enlazada — O(1)).

````python
from time import time
from collections import deque

for n in [2000, 4000, 8000, 16000]:
    lista = list(range(n))
    a = time()
    while lista:
        lista.pop(0)
    b = time()

    dq = deque(range(n))
    c = time()
    while dq:
        dq.popleft()
    d = time()

    print(f'n = {n:6d}   list.pop(0): {b-a:.4f}s   deque.popleft(): {d-c:.6f}s')
````
