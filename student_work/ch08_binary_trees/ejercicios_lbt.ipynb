## 2. Ejercicio 1 — ¿El árbol es completo?

Un árbol binario de altura $h$ es **completo** si

1. los niveles $0, 1, \dots, h-1$ tienen **todos** los nodos posibles ($2^i$ nodos en el nivel $i$), y
2. los nodos del último nivel $h$ están **lo más a la izquierda posible**: recorriendo ese nivel de izquierda a derecha no hay un hueco seguido de un nodo.

Es exactamente la forma que tiene un *heap* (cap. 9). El árbol vacío y el árbol de un solo nodo son completos.

**Ejemplos**

```
    A              A              A              A              A
   / \            / \            / \            / \              \
  B   C          B   C          B   C          B   C              C
 / \            / \   \        /              / \
D   E          D   E   G      D              D   E
                                            /
                                           H
completo       NO completo    completo       NO completo    NO completo
               (hueco antes   (último nivel  (el nivel 2
               de G)          a la izq.)     no está lleno)
```

Escribe `es_completo(T)` que retorne `True` si el `LinkedBinaryTree` `T` es completo y `False` en caso contrario.

*Pista:* piensa en un recorrido por niveles (*breadth-first*). ¿Qué pasa en un árbol completo después de encontrar el primer hijo que falta?

````python
# Árbol 1 — completo
#      A
#     / \
#    B   C
#   / \
#  D   E
T1 = LinkedBinaryTree()
a = T1._add_root('A')
b = T1._add_left(a, 'B')
c = T1._add_right(a, 'C')
d = T1._add_left(b, 'D')
e = T1._add_right(b, 'E')
````

````python
# Árbol 2 — NO completo (hueco antes de G)
#      A
#     / \
#    B   C
#   / \   \
#  D   E   G
T2 = LinkedBinaryTree()
a = T2._add_root('A')
b = T2._add_left(a, 'B')
c = T2._add_right(a, 'C')
d = T2._add_left(b, 'D')
e = T2._add_right(b, 'E')
g = T2._add_right(c, 'G')
````

````python
# Árbol 3 — completo (último nivel a la izquierda)
#      A
#     / \
#    B   C
#   /
#  D
T3 = LinkedBinaryTree()
a = T3._add_root('A')
b = T3._add_left(a, 'B')
c = T3._add_right(a, 'C')
d = T3._add_left(b, 'D')
````

````python
# Árbol 4 — NO completo (el nivel 2 no está lleno y ya hay nodos en el nivel 3)
#        A
#       / \
#      B   C
#     / \
#    D   E
#   /
#  H
T4 = LinkedBinaryTree()
a = T4._add_root('A')
b = T4._add_left(a, 'B')
c = T4._add_right(a, 'C')
d = T4._add_left(b, 'D')
e = T4._add_right(b, 'E')
h = T4._add_left(d, 'H')
````

## 3. Ejercicio 2 — El camino entre dos nodos

En un árbol hay **un único camino** entre dos nodos cualesquiera `p` y `q`: se sube desde `p` hasta el **ancestro común más bajo** de los dos y luego se baja hasta `q`.

Escribe `camino(T, p, q)` que retorne un **string** con los elementos de los nodos del camino, **desde `p` hasta `q`**, separados por `" -> "`.

Con el árbol que se construye abajo:

| `p` | `q` | `camino(T, p, q)` |
|---|---|---|
| `H` | `E` | `"H -> D -> B -> E"` |
| `H` | `G` | `"H -> D -> B -> A -> C -> G"` |
| `A` | `F` | `"A -> C -> F"` |
| `F` | `A` | `"F -> C -> A"` |
| `B` | `H` | `"B -> D -> H"` |
| `D` | `D` | `"D"` |

- El orden importa: `camino(T, p, q)` y `camino(T, q, p)` son uno el reverso del otro.
- Los elementos pueden no ser strings (por ejemplo enteros): usa `str(p.element())`.
- `p` y `q` son `Position` de `T` (no elementos).

El árbol de los ejemplos:

```
         A
      /     \
     B       C
    / \     / \
   D   E   F   G
  /
 H
```

````python
T = LinkedBinaryTree()
a = T._add_root('A')
b = T._add_left(a, 'B')
c = T._add_right(a, 'C')
d = T._add_left(b, 'D')
e = T._add_right(b, 'E')
f = T._add_left(c, 'F')
g = T._add_right(c, 'G')
h = T._add_left(d, 'H')
````

````python
ejemplos_camino = [
    (h, e, 'H -> D -> B -> E'),
    (h, g, 'H -> D -> B -> A -> C -> G'),
    (a, f, 'A -> C -> F'),
    (f, a, 'F -> C -> A'),
    (b, h, 'B -> D -> H'),
    (d, d, 'D'),
]
````

````python
%%writefile ejercicios_lbt.py
from goodrich.ch08.linked_binary_tree import LinkedBinaryTree


def es_completo(T):
    if T.is_empty():
        return True
    
    cola = [T.root()]
    encontro_hueco = False
    
    while cola:
        nodo = cola.pop(0)
        
        izq = T.left(nodo)
        if izq:
            if encontro_hueco:
                return False
            cola.append(izq)
        else:
            encontro_hueco = True
            
        der = T.right(nodo)
        if der:
            if encontro_hueco:
                return False
            cola.append(der)
        else:
            encontro_hueco = True
            
    return True


def camino(T, p, q):
    ancestros_p = []
    actual = p
    while True:
        ancestros_p.append(actual)
        if T.is_root(actual):
            break
        actual = T.parent(actual)
        
    ancestros_q = []
    actual = q
    while True:
        ancestros_q.append(actual)
        if T.is_root(actual):
            break
        actual = T.parent(actual)
        
    lca = None
    for nodo_p in ancestros_p:
        if nodo_p in ancestros_q:
            lca = nodo_p
            break
            
    idx_p = ancestros_p.index(lca)
    sub_p = ancestros_p[:idx_p + 1]
    
    idx_q = ancestros_q.index(lca)
    sub_q = ancestros_q[:idx_q]
    sub_q.reverse()
    
    camino_nodos = sub_p + sub_q
    
    elementos = [str(nodo.element()) for nodo in camino_nodos]
    return " -> ".join(elementos)


if __name__ == "__main__":
    pass
````
