```python
class StackVaciable(ArrayStack):
    """ArrayStack con un metodo adicional para vaciarse recursivamente."""

    def vaciar_recursivo(self):
        """Remueve recursivamente todos los elementos de la pila (sin bucles)."""
        if not self.is_empty():
            self.pop()
            self.vaciar_recursivo()
```

```python
def invertir_pila(S):
    """Reemplaza el contenido de S por los mismos elementos en orden invertido,
    usando unicamente transfer() y dos pilas temporales."""
    T1 = ArrayStack()
    T2 = ArrayStack()
    transfer(S, T1)
    transfer(T1, T2)
    transfer(T2, S)
```

```python
def evaluar_postfija(expr):
    """Evalua una expresion en notacion postfija (tokens separados por espacios)
    de forma no recursiva, usando una ArrayStack."""
    S = ArrayStack()
    for token in expr.split():
        if token in ("+", "-", "*", "/"):
            op2 = S.pop()
            op1 = S.pop()
            if token == "+":
                res = op1 + op2
            elif token == "-":
                res = op1 - op2
            elif token == "*":
                res = op1 * op2
            elif token == "/":
                res = op1 / op2
            S.push(res)
        else:
            S.push(float(token))
    return S.pop()
```

```python
class QueueDosPilas:
    def __init__(self):
        self._entrada = ArrayStack()
        self._salida = ArrayStack()

    def __len__(self):
        return len(self._entrada) + len(self._salida)

    def is_empty(self):
        return len(self) == 0

    def enqueue(self, e):
        self._entrada.push(e)

    def _trasvasar_si_hace_falta(self):
        """Si _salida esta vacia, mueve todo el contenido de _entrada a _salida."""
        if self._salida.is_empty():
            transfer(self._entrada, self._salida)

    def first(self):
        if self.is_empty():
            raise Empty('La cola está vacía')
        self._trasvasar_si_hace_falta()
        return self._salida.top()

    def dequeue(self):
        if self.is_empty():
            raise Empty('La cola está vacía')
        self._trasvasar_si_hace_falta()
        return self._salida.pop()
```

```python
# Respuesta:
# Peor caso de un dequeue individual: O(n)
# Caso amortizado sobre una secuencia de n operaciones: O(1)
# Justificacion:
# En el peor caso, si _salida está vacía, se deben mover n elementos de _entrada a _salida en O(n).
# Sin embargo, cada elemento entra a _entrada una vez (push), pasa a _salida una vez (pop y push) 
# y sale de _salida una vez (pop). Dado que cada elemento es manipulado exactamente un número constante 
# de veces a lo largo de su ciclo de vida, el costo total para n elementos es O(n), lo que da un costo 
# amortizado de O(1) por operación.
```
